```python
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer
import numpy as np

# Load the dataset (assuming your data is in a CSV named 'products.csv')
# NOTE: Replace 'products.csv' with your actual file path
df = pd.read_csv('/content/cleaned_flipkart_data.csv')

# --- 1. Filter Categories (Count > 50) ---
CATEGORY_COUNT_THRESHOLD = 50
category_counts = df['main_category'].value_counts()
top_categories = category_counts[category_counts > CATEGORY_COUNT_THRESHOLD].index.tolist()

# Filter the DataFrame to include only products from top categories
df_filtered = df[df['main_category'].isin(top_categories)].copy()
print(f"Total top categories used: {len(top_categories)}")
print(f"Filtered dataset size: {len(df_filtered)}")

# --- 2. Create Input Text (X) ---
# Combine key text features to create a rich input text string for the model
# Fill NaN values with an empty string before concatenating
df_filtered['product_name'] = df_filtered['product_name'].fillna('')
df_filtered['description'] = df_filtered['description'].fillna('')
df_filtered['product_specifications'] = df_filtered['product_specifications'].fillna('')

df_filtered['input_text'] = (
    df_filtered['product_name'] + " " +
    df_filtered['description'].str[:300] + " " + # Use a truncated description
    df_filtered['product_specifications'].str[:300]
)

# --- 3. Create Output Labels (Y) ---
# We need to treat this as Multi-Label Classification
# Since each product only has ONE main_category, we need to manually enrich the data
# For now, let's stick to the single main_category for simplicity in this code block:
# We will create a list of lists, where each inner list contains one category.

# Create the list of category lists
df_filtered['category_list'] = df_filtered['main_category'].apply(lambda x: [x])

# Use MultiLabelBinarizer to create the binary Y matrix
mlb = MultiLabelBinarizer()
Y_labels = mlb.fit_transform(df_filtered['category_list'])
NUM_LABELS = Y_labels.shape[1]

print(f"Shape of Y_labels (samples x categories): {Y_labels.shape}")
```

```python
# --- 4. Create Synthetic Intent Data (Crucial Step) ---
# Create a few hand-labeled examples to bridge the gap between product text and user intent
# The keys must be valid categories in your top_categories list.

# NOTE: You MUST create many more of these intent-to-category pairs for training!
synthetic_data = [
    {"intent": "I'm going on a trek and need gear", "categories": ["Footwear", "Clothing", "Bags, Wall..."]},
    {"intent": "Getting ready for a long hike", "categories": ["Footwear", "Clothing", "Sports & Fi..."]},
    {"intent": "Need something for my kitchen renovation", "categories": ["Kitchen & I...", "Home Dec...", "Tools & Ha..."]},
    {"intent": "Looking for accessories for my car", "categories": ["Automotiv", "Home Dec..."]},
    {"intent": "Need a gift for a new baby", "categories": ["Baby Care", "Toys & Sch..."]},
    {"intent": "I want to upgrade my PC", "categories": ["Computer...", "Watches"]}, # Watch is an example of an unrelated category
    {"intent": "I'm looking for a nice ring", "categories": ["Jewellery"]},
]

# Convert synthetic data to a DataFrame
synth_df = pd.DataFrame([
    {'input_text': item['intent'], 'category_list': item['categories']} for item in synthetic_data
])

# Use MLB to encode the synthetic labels
Y_synth = mlb.transform(synth_df['category_list'])

# Combine the real and synthetic data
X_train_text = pd.concat([df_filtered['input_text'], synth_df['input_text']], ignore_index=True)
Y_train = np.concatenate([Y_labels, Y_synth], axis=0)

print(f"Combined training size: {len(X_train_text)}")
```

```python
from sentence_transformers import SentenceTransformer
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split

# --- 5. Sentence Embeddings (Feature Engineering) ---
# Using a good pre-trained model like all-MiniLM-L6-v2 for speed and quality
model_name = 'all-MiniLM-L6-v2'
embedding_model = SentenceTransformer(model_name)
EMBEDDING_DIM = embedding_model.get_sentence_embedding_dimension()

print("Generating embeddings for all training text...")
# Encode the combined text data
X_train_embeddings = embedding_model.encode(X_train_text.tolist(), convert_to_tensor=True, show_progress_bar=True)
print(f"Embeddings shape (samples x dimension): {X_train_embeddings.shape}")

# Split the data for training and validation
X_train_tensor, X_val_tensor, Y_train_tensor, Y_val_tensor = train_test_split(
    X_train_embeddings, torch.tensor(Y_train, dtype=torch.float32), test_size=0.1, random_state=42
)

# --- 6. Define the Classification Model ---
class IntentClassifier(nn.Module):
    def __init__(self, input_dim, num_labels):
        super(IntentClassifier, self).__init__()
        self.classifier = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, num_labels) # Output layer with one logit per category
        )

    def forward(self, x):
        return self.classifier(x) # Output logits for Binary Cross-Entropy with Logits

# Initialize the model, loss, and optimizer
model = IntentClassifier(EMBEDDING_DIM, NUM_LABELS)
# Use BCEWithLogitsLoss for multi-label classification
criterion = nn.BCEWithLogitsLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

# Move model to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# --- 7. Training Loop (Simplified) ---
NUM_EPOCHS = 10
BATCH_SIZE = 32

train_dataset = torch.utils.data.TensorDataset(X_train_tensor, Y_train_tensor)
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)

print("\nStarting Model Training...")
for epoch in range(NUM_EPOCHS):
    model.train()
    total_loss = 0
    for batch_x, batch_y in train_loader:
        # Move data to GPU
        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad()
        logits = model(batch_x)
        loss = criterion(logits, batch_y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)
    print(f"Epoch {epoch+1}/{NUM_EPOCHS}, Loss: {avg_loss:.4f}")
```

```python
def predict_categories(user_query, embedding_model, classifier_model, mlb_binarizer, threshold=0.5, k=4):
    """
    Predicts the top K categories for a user query.
    """
    classifier_model.eval()

    # 1. Embed the user query
    query_embedding = embedding_model.encode([user_query], convert_to_tensor=True)

    # 2. Get the classification logits
    with torch.no_grad():
        logits = classifier_model(query_embedding)
        # Apply Sigmoid to convert logits to probabilities (0 to 1)
        probabilities = torch.sigmoid(logits).cpu().numpy()[0]

    # 3. Get the predicted categories

    # Method A: Binary prediction based on a fixed threshold (e.g., 0.5)
    # predicted_labels = (probabilities >= threshold).astype(int)
    # predicted_categories = mlb_binarizer.inverse_transform(predicted_labels[None, :])[0]

    # Method B: Rank and select top K categories (Better for recommendations)
    # Get indices of top K probabilities
    top_k_indices = np.argsort(probabilities)[::-1][:k]

    # Get the category names and their scores
    predicted_categories_ranked = [mlb_binarizer.classes_[i] for i in top_k_indices]
    scores = [probabilities[i] for i in top_k_indices]

    return predicted_categories_ranked, scores

# --- Test the System ---
print("\n--- Testing the Intent-to-Category System ---")
USER_INPUT = "I’m going on a trek"

recommended_categories, scores = predict_categories(USER_INPUT, embedding_model, model, mlb, k=4)

print(f"User Query: '{USER_INPUT}'")
print("\nRecommended Categories:")
for cat, score in zip(recommended_categories, scores):
    print(f"- **{cat}** (Confidence: {score:.2f})")

# Example Output should look like:
# - **Footwear** (Confidence: 0.95)
# - **Clothing** (Confidence: 0.92)
# - **Bags, Wall...** (Confidence: 0.88)
# - **Sports & Fi...** (Confidence: 0.75)
```

```python
# --- 1. Create Product Embeddings (if not already done) ---
# If you've already created embeddings for the 'input_text' in a previous step,
# you might be able to reuse them. Otherwise, generate them here.

# Assuming 'df' is your full product DataFrame and 'input_text' column exists
# (created by combining product_name, description, specifications)
# Ensure 'input_text' column is created if not already present
if 'input_text' not in df.columns:
    df['input_text'] = (
        df['product_name'].fillna('') + " " +
        df['description'].fillna('').str[:300] + " " +
        df['product_specifications'].fillna('').str[:300]
    )

# Use the same embedding model trained earlier
# embedding_model = SentenceTransformer('all-MiniLM-L6-v2') # Assuming this is loaded

print("Generating embeddings for all product texts...")
# Generate embeddings for the product input text
# This can take some time for a large dataset
product_embeddings = embedding_model.encode(df['input_text'].tolist(), convert_to_tensor=True, show_progress_bar=True)
print(f"Product embeddings shape: {product_embeddings.shape}")


# --- 2. Implement Semantic Search Function ---
from sentence_transformers.util import semantic_search

def semantic_product_search(user_query, embedding_model, product_embeddings, products_df, top_k=10):
    """
    Performs semantic search to find top K most relevant products for a user query.
    """
    # Embed the user query
    query_embedding = embedding_model.encode(user_query, convert_to_tensor=True)

    # Perform semantic search
    # Returns a list of dictionaries, one for each query.
    # Each dictionary contains a list of corpus ids and their similarity scores.
    hits = semantic_search(query_embedding, product_embeddings, top_k=top_k)

    # Get the product indices and scores
    # Assuming a single user query, so take the first element of the hits list
    hit_list = hits[0]

    # Retrieve the actual product information using the indices
    recommended_products = []
    for hit in hit_list:
        corpus_id = hit['corpus_id']
        score = hit['score']
        product_info = products_df.iloc[corpus_id].copy() # Get the product row
        product_info['similarity_score'] = score # Add the similarity score
        recommended_products.append(product_info)

    return pd.DataFrame(recommended_products)

# --- 3. Test the Semantic Search ---
print("\n--- Testing the Semantic Search System ---")
USER_INPUT_SEMANTIC = "I’m going on a trek" # Example broad query

# Perform the semantic search
recommended_products_semantic = semantic_product_search(
    USER_INPUT_SEMANTIC,
    embedding_model,
    product_embeddings,
    df, # Use the original dataframe to get full product info
    top_k=10
)

# Display the results
print(f"User Query: '{USER_INPUT_SEMANTIC}'")
print("\nTop 10 Recommended Products (Semantic Search):")
display(recommended_products_semantic[['product_name', 'main_category', 'similarity_score']])
```

## Define graph state

### Subtask:
Create a `TypedDict` to define the structure of the information that will be passed between nodes in the LangGraph.


**Reasoning**:
Define a `TypedDict` to represent the state of the LangGraph, including fields for the user query and semantic search results.



```python
from typing import TypedDict, List, Dict, Any

class GraphState(TypedDict):
    """
    Represents the state of our graph.

    Attributes:
        user_query: user query
        semantic_search_results: results from the semantic search node
    """
    user_query: str
    semantic_search_results: List[Dict[str, Any]]
```

## Define nodes

### Subtask:
Create Python functions for each step of the workflow (e.g., a node for semantic search).


**Reasoning**:
Define the `semantic_search_node` function to perform semantic search and update the graph state.



```python
def semantic_search_node(state: GraphState) -> GraphState:
    """
    Performs semantic search based on the user query and updates the state.

    Args:
        state: The current graph state.

    Returns:
        The updated graph state with semantic search results.
    """
    print("---SEMANTIC SEARCH NODE---")
    user_query = state.get("user_query")
    print(f"Searching for: {user_query}")

    if not user_query:
        print("No user query found in state. Skipping semantic search.")
        state["semantic_search_results"] = []
        return state

    # Assuming semantic_product_search, embedding_model, product_embeddings, and df are available
    # from previous cells in the notebook environment.
    try:
        recommended_products_df = semantic_product_search(
            user_query,
            embedding_model,
            product_embeddings,
            df, # Use the original dataframe
            top_k=10 # Get the top 10 results
        )
        # Convert DataFrame to a list of dictionaries for the state
        state["semantic_search_results"] = recommended_products_df.to_dict('records')
        print(f"Found {len(state['semantic_search_results'])} semantic search results.")
    except Exception as e:
        print(f"Error during semantic search: {e}")
        state["semantic_search_results"] = [] # Ensure the key exists even on error

    return state
```

## Build the graph

### Subtask:
Instantiate a `StateGraph` and add the defined nodes and edges, including a starting point.


**Reasoning**:
Instantiate the StateGraph and add the semantic search node as the entry point.



```python
from langgraph.graph import StateGraph

# Instantiate the StateGraph
workflow = StateGraph(GraphState)

# Add the semantic search node
workflow.add_node("semantic_search", semantic_search_node)

# Set the entry point
workflow.set_entry_point("semantic_search")
```

## Compile and run

### Subtask:
Compile the graph and demonstrate how to invoke it with a user query.


**Reasoning**:
Compile the defined workflow and invoke it with a user query to demonstrate its functionality.



```python
# Compile the graph
app = workflow.compile()

# Define a user query
user_query = "looking for a sturdy hiking backpack"

# Invoke the compiled graph
print(f"Invoking graph with query: '{user_query}'")
output = app.invoke({"user_query": user_query})

# Print the output
print("\n--- Graph Execution Output ---")
# Check if semantic search results are available in the output
if "semantic_search_results" in output and output["semantic_search_results"]:
    print("Recommended Products:")
    # Iterate through the list of dictionaries and print relevant information
    for product in output["semantic_search_results"]:
        product_name = product.get("product_name", "N/A")
        main_category = product.get("main_category", "N/A")
        similarity_score = product.get("similarity_score", "N/A")
        print(f"- Product: {product_name}")
        print(f"  Category: {main_category}")
        print(f"  Similarity Score: {similarity_score:.4f}" if isinstance(similarity_score, float) else f"  Similarity Score: {similarity_score}")
        print("-" * 20) # Separator for readability
else:
    print("No semantic search results found in the output.")
```
